## 2. Configuration


```python
# Configuration (update these with your values)
GENIE_SPACE_ID = "01f0b223e0e31cb0b4b093ef8578bcf8"
CATALOG = "humana_quality"
SCHEMA = "hedis_gold"
KNOWLEDGE_ASSISTANT_ENDPOINT_ID = "ka-0bbaf97e-endpoint"  # Optional
AI_MODEL_NAME = "databricks-meta-llama-3-1-8b-instruct"

print(f"📋 Configuration:")
print(f"   Genie Space: {GENIE_SPACE_ID}")
print(f"   Catalog: {CATALOG}")
print(f"   Schema: {SCHEMA}")
print(f"   AI Model: {AI_MODEL_NAME}")
```

## 3. Initialize HEDIS Agent

**This imports the same agent used in production!**


```python
# Import the HEDIS Agent
from hedis_agent import HEDISQualityAgent

print("✅ HEDISQualityAgent imported successfully")
```

```python
# Initialize the agent
print("🤖 Initializing HEDIS MCP Agent...")
print("   This may take 10-20 seconds...\\n")

agent = HEDISQualityAgent(
    genie_space_id=GENIE_SPACE_ID,
    catalog=CATALOG,
    schema=SCHEMA,
    knowledge_assistant_endpoint_id=KNOWLEDGE_ASSISTANT_ENDPOINT_ID,
    ai_model_name=AI_MODEL_NAME
)

print("\\n✅ Agent initialized successfully!")
```

## 4. Check System Status

Verify all MCP components are working


```python
# Get system status
status = agent.get_system_status()

print("🔍 System Status:")
print(f"\\n1️⃣  LLM Ready: {'✅ Yes' if status['llm_ready'] else '❌ No'}")
print(f"2️⃣  Total Tools: {status['tools_count']}")

print(f"\\n3️⃣  Genie MCP:")
genie = status['genie']
print(f"   Status: {genie.get('status', 'unknown')}")
print(f"   Tools: {genie.get('tools_count', 0)}")

print(f"\\n4️⃣  UC Functions MCP:")
uc = status['uc_functions']
print(f"   Status: {uc.get('status', 'unknown')}")
print(f"   Tools: {uc.get('tools_count', 0)}")
if uc.get('tools'):
    print(f"   Available functions:")
    for tool in uc['tools']:
        print(f"      - {tool}")

print(f"\\n5️⃣  Knowledge Assistant:")
ka = status['knowledge_assistant']
print(f"   Status: {ka.get('status', 'unknown')}")

print("\\n" + "="*60)
```

## 5. Test Member Queries

Test UC Functions for member lookups


```python
# Test member query
query = "Show me member M000001's quality metrics"
print(f"💬 Query: {query}\\n")
print("🤖 Agent is thinking...\\n")

response = agent.chat(query, [])

print("📄 Response:")
print("="*60)
print(response)
print("="*60)
```
